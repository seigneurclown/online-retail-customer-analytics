## 0. Chuẩn bị thư viện & Nạp dữ liệu cấp hóa đơn

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
from scipy import stats

current_dir = Path.cwd().resolve()
PROJECT_ROOT = current_dir.parent if current_dir.name == "notebooks" else current_dir
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.analysis.statistics import (
    test_weekday_vs_weekend_invoice_value,
    test_country_vs_cancellation,
)

pd.options.display.float_format = "{:.4f}".format

invoice_df = pd.read_csv(PROJECT_ROOT / "data" / "processed" / "invoice_data.csv")
print(f"Nạp thành công {len(invoice_df):,} hóa đơn từ invoice_data.csv")
```

## 1. Kiểm định 1: Giá trị đơn hàng giữa Weekday và Weekend

- **Research Question:** Có sự khác biệt có ý nghĩa thống kê về giá trị đơn hàng giữa ngày trong tuần và ngày cuối tuần không?
- **H0:** Phân phối giá trị đơn hàng giữa Weekday và Weekend là như nhau.
- **H1:** Phân phối giá trị đơn hàng giữa Weekday và Weekend khác nhau.
- **Lý do chọn Mann-Whitney U test:** Dữ liệu hóa đơn lệch phải cực mạnh (vi phạm giả định chuẩn của T-test).

```python
res1 = test_weekday_vs_weekend_invoice_value(invoice_df)

print(f"Kiểm định sử dụng: {res1['TestName']}")
print(f"Trung vị Weekday  : £{res1['WeekdayMedian']:,.2f} (n = {res1['WeekdayCount']:,})")
print(f"Trung vị Weekend  : £{res1['WeekendMedian']:,.2f} (n = {res1['WeekendCount']:,})")
print(f"Thống kê U       : {res1['Statistic']:,}")
print(f"p-value          : {res1['p_value']:.4e}")
print(f"Bác bỏ H0        : {res1['Reject_H0']}")
print(f"\nKết luận:\n{res1['Interpretation']}")
print(f"\nGóc nhìn kinh doanh:\n{res1['BusinessInsight']}")
print(f"\nHạn chế dữ liệu:\n{res1['DataLimitation']}")
```

## 2. Kiểm định 2: Thị trường Quốc gia và Tỷ lệ hủy đơn (Cancellation)

- **Research Question:** Thị trường quốc gia có liên quan đến hành vi hủy đơn hàng hay không?
- **H0:** Quốc gia và trạng thái hủy đơn độc lập với nhau.
- **H1:** Quốc gia và trạng thái hủy đơn có mối liên hệ phụ thuộc.
- **Lý do chọn Chi-Square Test of Independence:** Kiểm định mối liên hệ giữa 2 biến định tính (Categorical variables) với tần số kỳ vọng mỗi ô đều >= 5.

```python
res2 = test_country_vs_cancellation(invoice_df)

print(f"Kiểm định sử dụng   : {res2['TestName']}")
print(f"Bậc tự do (df)      : {res2['DegreesOfFreedom']}")
print(f"Thống kê Chi2 (χ²)  : {res2['Statistic']:.3f}")
print(f"p-value             : {res2['p_value']:.4e}")
print(f"Bác bỏ H0           : {res2['Reject_H0']}")

print("\nTỷ lệ hủy đơn (%): ")
for c, rate in res2["CancellationRates"].items():
    print(f"  - {c:16s}: {rate:.2f}%")

print(f"\nKết luận:\n{res2['Interpretation']}")
print(f"\nGóc nhìn kinh doanh:\n{res2['BusinessInsight']}")
```
