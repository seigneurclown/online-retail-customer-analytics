1. Business Requirements 

Phân tích tình hình và xu hướng doanh thu.  

Phân tích hành vi mua hàng của khách hàng.  

Nhận diện các nhóm khách hàng có hành vi tương đồng.  

Cung cấp insight hỗ trợ quyết định kinh doanh.  

2. Data Requirements 

Dữ liệu giao dịch: InvoiceNo, StockCode, Quantity, InvoiceDate, UnitPrice, CustomerID, Country.  

Kiểm tra dữ liệu thiếu, trùng lặp và không hợp lệ.  

Chuẩn hóa dữ liệu trước khi phân tích.  

3. Analytics Requirements 

Tính doanh thu và phân tích theo thời gian, sản phẩm, thị trường.  

Trực quan hóa bằng biểu đồ phù hợp.  

Thực hiện thống kê mô tả và kiểm định giả thuyết.  

Xây dựng RFM để mô tả hành vi khách hàng.  

4. Machine Learning Requirements 

Sử dụng K-Means Clustering để phân cụm khách hàng.  

Thử và lựa chọn số cụm phù hợp.  

Đánh giá bằng Elbow Method + Silhouette Score.  

Phân tích đặc điểm của từng cụm.  

5. Output / Business Insights 

Báo cáo doanh thu và xu hướng.  

Hồ sơ từng nhóm khách hàng.  

Các insight từ kết quả phân tích.  

Đề xuất hướng tiếp cận khách hàng dựa trên dữ liệu.  

6. Validation 

Kiểm tra chất lượng dữ liệu sau cleaning.  

Kiểm tra tính đúng của doanh thu và RFM.  

Đánh giá chất lượng phân cụm.  

Kiểm tra khả năng giải thích kết quả theo góc nhìn doanh nghiệp.
